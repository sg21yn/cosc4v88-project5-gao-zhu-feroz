## Interpretation
- The flag sends questionable **explanations** for **human review**; it does not revise an applicant's credit score.
- The outcome is a separate deletion-test proxy (LIME top three features vs three random features), not objective ground truth.
- Logistic Regression: about 30% review catches 29 of 38 deletion-test failures.
- XGBoost: the four-signal flag catches 25 of 67. At the same 77 reviews, the borderline signal alone catches 42 (about 17 expected by chance), so for XGBoost LIME instability is not a good proxy for unfaithfulness.
- Deletion-test failures cluster near the decision threshold, so part of the borderline advantage may come from the test itself.
- The workload-matched comparison uses the evaluation review count, so it is a check, not a deployable rule.
- **Deployment recommendation:** no real-world deployment; research/demo only.
